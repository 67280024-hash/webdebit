```python
import pandas as pd
import numpy as np

messy_data = {
    'name': ['Alice', 'Bob', None, 'David', 'Eve', 'Frank', 'Bob'],
    'age': [25, 30, 35, None, 32, 28, 30],
    'city': ['Bangkok', None, 'Phuket', 'Bangkok', 'Pattaya',
             'Bangkok', 'Phuket'],
    'salary': [50000, 60000, np.nan, 55000, 65000, 52000, 60000],
    'email': ['alice@email.com', 'BOB@EMAIL.COM', 'charlie@email.com',
              'david@email.com', 'eve@email.com', 'frank@email.com',
              'bob@email.com']
}
df_messy = pd.DataFrame(messy_data)
df_clean = df_messy.dropna(subset=['name']).copy()
df_clean['age'] = df_clean['age'].fillna(df_clean['age'].mean())
df_clean['city'] = df_clean['city'].fillna('Unknown')
df_clean['salary'] = df_clean['salary'].fillna(df_clean['salary'].median())
print(df_clean)
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
months = np.arange(1, 13)
sales_A = np.linspace(100, 200, 12) + np.random.randint(-10, 10, 12)
sales_B = np.linspace(80, 180, 12) + np.random.randint(-10, 10, 12)
sales_C = np.linspace(60, 160, 12) + np.random.randint(-10, 10, 12)
bar_width = 0.25
x = np.arange(len(months))
plt.figure(figsize=(12, 6))
colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
bars_A = plt.bar(x - bar_width, sales_A, width=bar_width, label='Product A', color=colors[0])
bars_B = plt.bar(x, sales_B, width=bar_width, label='Product B', color=colors[1])
bars_C = plt.bar(x + bar_width, sales_C, width=bar_width, label='Product C', color=colors[2])
plt.xticks(x, [f'M{m}' for m in months])
plt.xlabel('Month')
plt.ylabel('Sales')
plt.title('Monthly Sales Comparison (Products A, B, C)')
for bars in [bars_A, bars_B, bars_C]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height,
                 f'{int(height)}', ha='center', va='bottom', fontsize=8)
plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left')
plt.grid(axis='y')
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
all_sales = np.concatenate([sales_A, sales_B, sales_C])
max_value = np.max(all_sales)
if max_value in sales_A:
    idx = np.argmax(sales_A)
    x_pos = x[idx] - bar_width
elif max_value in sales_B:
    idx = np.argmax(sales_B)
    x_pos = x[idx]
else:
    idx = np.argmax(sales_C)
    x_pos = x[idx] + bar_width
plt.annotate('Highest Sales',
             xy=(x_pos, max_value),
             xytext=(x_pos, max_value + 20),
             arrowprops=dict(arrowstyle='->'))
plt.tight_layout()
plt.show()
```
