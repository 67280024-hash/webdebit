```python
import numpy as np
import pandas as pd

# สร้างข้อมูลที่มี outliers
np.random.seed(42)
data = np.random.normal(100, 15, 100)
data = np.append(data, [200, 250, 300])  # เพิ่ม outliers

df = pd.DataFrame({'value': data})

# -------------------------------
# 1. Z-score method
mean = df['value'].mean()
std = df['value'].std()
df['z_score'] = (df['value'] - mean) / std

# ตัดที่ |z| > 3
df_no_outliers_z = df[np.abs(df['z_score']) < 3]

# -------------------------------
# 2. IQR method
Q1 = df['value'].quantile(0.25)
Q3 = df['value'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_no_outliers_iqr = df[
    (df['value'] >= lower_bound) &
    (df['value'] <= upper_bound)
]

# -------------------------------
# 3. Percentile method
lower = df['value'].quantile(0.01)
upper = df['value'].quantile(0.99)

df_no_outliers_percentile = df[
    (df['value'] >= lower) &
    (df['value'] <= upper)
]

# -------------------------------
# 4. Capping (แทนที่ outliers ด้วยขอบเขต)
df['value_capped'] = df['value'].clip(
    lower=lower_bound,
    upper=upper_bound
)

print(df.head())
```
