```python
import pandas as pd

# สร้าง DataFrame
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'city': ['Bangkok', 'Chiang Mai', 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, 75000, 55000, 65000]
}

df = pd.DataFrame(data)

# ===============================
# เลือกคอลัมน์
# ===============================

# เลือกคอลัมน์เดียว (ได้ Series)
ages = df['age']
print(type(ages))  # pandas.core.series.Series

# เลือกหลายคอลัมน์ (ได้ DataFrame)
subset = df[['name', 'salary']]
print(type(subset))  # pandas.core.frame.DataFrame

# เลือกแบบ attribute
ages = df.age

# ===============================
# เลือกแถว
# ===============================

# Slicing by position
print(df[0:3])     # แถว 0-2
print(df[:2])      # 2 แถวแรก
print(df[-2:])     # 2 แถวสุดท้าย

# ===============================
# Boolean indexing
# ===============================
young = df[df['age'] < 30]
print(young)

bangkok_high_salary = df[
    (df['city'] == 'Bangkok') & (df['salary'] > 50000)
]
print(bangkok_high_salary)

# Multiple conditions
result = df[(df['age'] >= 30) | (df['salary'] > 70000)]
print(result)

# ===============================
# loc และ iloc
# ===============================

# loc - เลือกด้วย label
print(df.loc[0])                         # แถวที่ 0 (ทั้งแถว)
print(df.loc[0, 'name'])                 # แถว 0, คอลัมน์ name
print(df.loc[0:2, 'name'])               # แถว 0-2, คอลัมน์ name
print(df.loc[0:2, ['name', 'age']])      # แถว 0-2, หลายคอลัมน์
print(df.loc[:, 'name':'city'])          # ทุกแถว, name ถึง city

# iloc - เลือกด้วย position
print(df.iloc[0])                        # แถวที่ 0
print(df.iloc[0, 0])                     # แถว 0, คอลัมน์ 0
print(df.iloc[0:2, 0:2])                 # แถว 0-1, คอลัมน์ 0-1
print(df.iloc[:, 0])                     # ทุกแถว, คอลัมน์แรก

# ===============================
# Boolean indexing with loc (แก้ไขค่า)
# ===============================
df.loc[df['age'] > 30, 'salary'] = (
    df.loc[df['age'] > 30, 'salary'] * 1.1
)

print(df)
```
