```python
import pandas as pd

# ===============================
# ข้อมูลตั้งต้น
# ===============================
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'salary': [50000, 60000, 75000],
    'city': ['Bangkok', 'Bangkok', 'Bangkok']
})

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# เปลี่ยนชื่อคอลัมน์บางคอลัมน์
# ===============================
df = df.rename(columns={
    'name': 'employee_name',
    'age': 'employee_age'
})

print("\nหลังเปลี่ยนชื่อบางคอลัมน์")
print(df)

# ===============================
# เปลี่ยนหลายคอลัมน์อีกครั้ง
# ===============================
df = df.rename(columns={
    'employee_name': 'name',
    'employee_age': 'age',
    'salary': 'monthly_salary'
})

print("\nหลังเปลี่ยนหลายคอลัมน์")
print(df)

# ===============================
# ตั้งชื่อคอลัมน์ใหม่ทั้งหมด
# ===============================
df.columns = ['Name', 'Age', 'Salary', 'City']

print("\nตั้งชื่อคอลัมน์ใหม่ทั้งหมด")
print(df)

# ===============================
# แปลงชื่อคอลัมน์เป็นตัวพิมพ์เล็ก
# ===============================
df.columns = df.columns.str.lower()

print("\nชื่อคอลัมน์เป็นตัวพิมพ์เล็ก")
print(df)

# ===============================
# เปลี่ยนชื่อ index
# ===============================
df = df.rename(index={0: 'first', 1: 'second'})

print("\nหลังเปลี่ยนชื่อ index")
print(df)

# ===============================
# Reset index
# ===============================
df = df.reset_index(drop=True)

print("\nหลัง reset index")
print(df)

# ===============================
# Set index
# ===============================
df = df.set_index('name')

print("\nตั้ง name เป็น index")
print(df)
```
