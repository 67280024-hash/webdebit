```python
import seaborn as sns
import matplotlib.pyplot as plt

# ตั้ง theme
sns.set_theme(style="whitegrid", palette="Set2")

# =========================
# 3.1 Titanic Dataset
# =========================
titanic = sns.load_dataset('titanic')

plt.figure(figsize=(15, 12))

# a) Count plot: survived by class
plt.subplot(3, 2, 1)
sns.countplot(data=titanic, x="class", hue="survived")
plt.title("Survival Count by Class")

# b) Count plot: survived by sex
plt.subplot(3, 2, 2)
sns.countplot(data=titanic, x="sex", hue="survived")
plt.title("Survival Count by Sex")

# c) Histogram: age distribution
plt.subplot(3, 2, 3)
sns.histplot(titanic["age"].dropna(), bins=30, kde=True)
plt.title("Age Distribution")

# d) Box plot: age by class
plt.subplot(3, 2, 4)
sns.boxplot(data=titanic, x="class", y="age")
plt.title("Age by Passenger Class")

# e) Bar plot: average fare by class
plt.subplot(3, 2, 5)
sns.barplot(
    data=titanic,
    x="class",
    y="fare",
    estimator="mean",
    errorbar=None
)
plt.title("Average Fare by Class")

plt.tight_layout()
plt.show()


# =========================
# 3.2 Tips Dataset
# =========================
tips = sns.load_dataset('tips')

plt.figure(figsize=(15, 12))

# a) Scatter plot: total_bill vs tip (hue=time)
plt.subplot(3, 2, 1)
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time")
plt.title("Total Bill vs Tip (Time)")

# b) Box plot: total_bill by day (hue=sex)
plt.subplot(3, 2, 2)
sns.boxplot(data=tips, x="day", y="total_bill", hue="sex")
plt.title("Total Bill by Day and Sex")

# c) Violin plot: tip by day (split by sex)
plt.subplot(3, 2, 3)
sns.violinplot(data=tips, x="day", y="tip", hue="sex", split=True)
plt.title("Tip Distribution by Day and Sex")

# d) Bar plot: average tip by day
plt.subplot(3, 2, 4)
sns.barplot(
    data=tips,
    x="day",
    y="tip",
    estimator="mean",
    errorbar=None
)
plt.title("Average Tip by Day")

# e) Histogram: total_bill distribution (hue=smoker)
plt.subplot(3, 2, 5)
sns.histplot(
    data=tips,
    x="total_bill",
    hue="smoker",
    bins=30,
    kde=True
)
plt.title("Total Bill Distribution by Smoker")

plt.tight_layout()
plt.show()
```
