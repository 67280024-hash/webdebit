```python
import pandas as pd
import numpy as np

temperatures = pd.Series([32, 34, 31, 29, 30])
print(temperatures)

temperatures = pd.Series(
    [32, 34, 31, 29, 30],
    index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
)
print(temperatures)

data = {'Mon': 32, 'Tue': 34, 'Wed': 31, 'Thu': 29, 'Fri': 30}
temperatures = pd.Series(data)
print(temperatures)

arr = np.array([32, 34, 31, 29, 30])
temperatures = pd.Series(arr, index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])
print(temperatures)
```
