```python
import pandas as pd
import numpy as np

# ===============================
# สร้าง DataFrame (มี missing values)
# ===============================
data = {
    'name': ['Alice', 'Bob', None, 'David', 'Eve'],
    'age': [25, 30, 35, None, 32],
    'city': ['Bangkok', None, 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, np.nan, 55000, 65000]
}

df = pd.DataFrame(data)
print("Before handling missing values:")
print(df)

# ===============================
# สร้าง missing indicator
# ===============================
df['age_missing'] = df['age'].isna()

# เติม missing ด้วยค่าพิเศษ (สำหรับ ML บางโมเดล)
df['age'] = df['age'].fillna(-999)

print("\nAfter handling missing values:")
print(df)
```
