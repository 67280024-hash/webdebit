```python
import pandas as pd

# ===============================
# ข้อมูลตัวอย่าง
# ===============================
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'salary': [50000, 55000, 60000, 65000, 45000, 48000],
    'age': [25, 30, 28, 35, 27, 32],
    'department': ['IT', 'HR', 'IT', 'HR', 'IT', 'HR'],
    'gender': ['F', 'M', 'M', 'M', 'F', 'M']
})

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# GroupBy ตาม department
# ===============================
group_dept = df.groupby('department')

print("\nเงินเดือนรวมแต่ละแผนก")
print(group_dept['salary'].sum())

print("\nเงินเดือนเฉลี่ยแต่ละแผนก")
print(group_dept['salary'].mean())

print("\nเงินเดือนสูงสุดแต่ละแผนก")
print(group_dept['salary'].max())

print("\nเงินเดือนต่ำสุดแต่ละแผนก")
print(group_dept['salary'].min())

print("\nจำนวนพนักงานแต่ละแผนก")
print(group_dept['salary'].count())

print("\nส่วนเบี่ยงเบนมาตรฐานของอายุแต่ละแผนก")
print(group_dept['age'].std())

# ===============================
# Aggregation หลายฟังก์ชัน
# ===============================
agg_result = group_dept.agg({
    'salary': ['mean', 'sum', 'max'],
    'age': ['mean', 'min', 'max']
})

print("\nAggregation หลายฟังก์ชัน")
print(agg_result)

# ===============================
# ค่าเฉลี่ย salary และ age
# ===============================
mean_result = group_dept[['salary', 'age']].mean()
print("\nค่าเฉลี่ย salary และ age")
print(mean_result)

# ===============================
# GroupBy หลายคอลัมน์ (department + gender)
# ===============================
group_dept_gender = df.groupby(['department', 'gender'])

def salary_range(x):
    return x.max() - x.min()

salary_range_result = group_dept_gender['salary'].agg(salary_range)
print("\nช่วงเงินเดือน (max - min) แยกตามแผนกและเพศ")
print(salary_range_result)

# ===============================
# Named Aggregation
# ===============================
named_agg = df.groupby('department').agg(
    avg_salary=('salary', 'mean'),
    max_salary=('salary', 'max'),
    count=('salary', 'count'),
    avg_age=('age', 'mean')
)

print("\nNamed Aggregation")
print(named_agg)

# ===============================
# หาพนักงานเงินเดือนสูงสุดของแต่ละแผนก
# ===============================
def get_highest_paid(group):
    return group.loc[group['salary'].idxmax()]

highest_paid = group_dept.apply(get_highest_paid)
print("\nพนักงานเงินเดือนสูงสุดของแต่ละแผนก")
print(highest_paid)

# ===============================
# Transform (คืนค่าขนาดเดิม)
# ===============================
df['dept_avg_salary'] = group_dept['salary'].transform('mean')
df['salary_vs_dept_avg'] = df['salary'] - df['dept_avg_salary']

print("\nDataFrame หลังใช้ transform")
print(df)

# ===============================
# Filter กลุ่มที่เงินเดือนเฉลี่ย > 50000
# ===============================
filtered_groups = group_dept.filter(lambda x: x['salary'].mean() > 50000)
print("\nกลุ่มที่เงินเดือนเฉลี่ยมากกว่า 50000")
print(filtered_groups)

# ===============================
# Size และ Value Counts
# ===============================
print("\nจำนวนสมาชิกแต่ละแผนก")
print(group_dept.size())

print("\nจำนวนเพศในแต่ละแผนก")
print(df.groupby('department')['gender'].value_counts())
```
