```python
import pandas as pd
import numpy as np

# ===============================
# สร้าง DataFrame ตัวอย่าง
# ===============================
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'salary': [50000, 60000, 75000, 55000, 65000],
    'city': ['Bangkok', 'Phuket', 'Bangkok', 'Pattaya', 'Phuket'],
    'date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04', '2024-01-05'],
    'value': ['10', '20', 'x', '40', '50']  # มีค่าที่แปลงไม่ได้
}

df = pd.DataFrame(data)

# ===============================
# ตรวจสอบ data types
# ===============================
print(df.dtypes)

# ===============================
# แปลง data type ทีละคอลัมน์
# ===============================
df['age'] = df['age'].astype(int)
df['salary'] = df['salary'].astype(float)
df['name'] = df['name'].astype(str)

# ===============================
# แปลงหลายคอลัมน์พร้อมกัน
# ===============================
df = df.astype({
    'age': int,
    'salary': float,
    'city': 'category'   # category ประหยัด memory
})

# ===============================
# แปลงเป็น datetime
# ===============================
df['date'] = pd.to_datetime(df['date'])
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d')

# ===============================
# แปลงเป็น numeric (แปลงไม่ได้ → NaN)
# ===============================
df['value'] = pd.to_numeric(df['value'], errors='coerce')

# ===============================
# Category type
# ===============================
df['city'] = df['city'].astype('category')
print(df['city'].cat.categories)

# แสดงผลลัพธ์สุดท้าย
print("\nFinal DataFrame:")
print(df)
print("\nFinal dtypes:")
print(df.dtypes)
```
