```python
import pandas as pd
import numpy as np

# Create Series
temperatures = pd.Series(
    [32, 34, 31, 29, 30],
    index=['Mon', 'Tue', 'Wed', 'Thu', 'Fri']
)

# Attributes
print(temperatures.values)     # array([32, 34, 31, 29, 30])
print(temperatures.index)      # Index(['Mon', 'Tue', 'Wed', 'Thu', 'Fri'])
print(temperatures.dtype)      # int64
print(temperatures.shape)      # (5,)
print(temperatures.size)       # 5

# Accessing data
print(temperatures['Mon'])     # 32
print(temperatures)            # แสดงทั้ง Series
print(temperatures[['Mon', 'Wed']])  # เลือกหลายค่า

# Slicing
print(temperatures['Mon':'Wed'])  # label-based (รวม Wed)
print(temperatures[0:3])          # position-based (0–2)

# Boolean indexing
hot_days = temperatures[temperatures > 31]
print(hot_days)

# Methods
print(temperatures.mean())     # 31.2
print(temperatures.max())      # 34
print(temperatures.min())      # 29
print(temperatures.std())      # ส่วนเบี่ยงเบนมาตรฐาน
print(temperatures.describe()) # สถิติสรุป

# Operations
print(temperatures + 5)            # บวกทุกค่า
print(temperatures * 1.8 + 32)     # แปลง Celsius -> Fahrenheit
print(temperatures.sort_values())  # เรียงตามค่า
print(temperatures.sort_index())   # เรียงตาม index
```
