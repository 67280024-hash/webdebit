```python
import pandas as pd

# สร้างข้อมูลตัวอย่าง
df = pd.DataFrame({
    'name': ['  alice  ', 'BOB', 'charlie', 'DAVID'],
    'email': ['alice@email.com', 'bob@EMAIL.COM',
              'charlie@email.com', 'david@EMAIL.COM'],
    'phone': ['081-234 5678', '082 345-6789',
              '083-456-7890', '084 567 8901'],
    'full_name': ['alice smith', 'bob jones',
                  'charlie brown', 'david lee']
})

# String methods (ใช้ .str)
df['name_clean'] = df['name'].str.strip()        # ตัด whitespace
df['name_lower'] = df['name'].str.lower()        # ตัวพิมพ์เล็ก
df['name_upper'] = df['name'].str.upper()        # ตัวพิมพ์ใหญ่
df['name_title'] = df['name'].str.title()        # Title Case
df['name_length'] = df['name'].str.len()         # ความยาว

# Replace
df['email_clean'] = df['email'].str.lower()
df['email_domain'] = df['email'].str.split('@').str[1]

# Contains
has_alice = df['name'].str.contains('alice', case=False)
print(df[has_alice])

# Extract with regex
df['username'] = df['email'].str.extract(r'(.+)@')

# Split
df[['first', 'last']] = df['full_name'].str.split(' ', expand=True)

# ตัวอย่างการทำความสะอาดข้อมูล
df['phone'] = df['phone'].str.replace('-', '', regex=False)
df['phone'] = df['phone'].str.replace(' ', '', regex=False)
df['phone'] = df['phone'].str.strip()

# Startswith, Endswith
gmail_users = df[df['email_clean'].str.endswith('@gmail.com')]

print(df)
```
