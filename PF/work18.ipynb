```python
import pandas as pd

# ===============================
# ข้อมูลตัวอย่าง
# ===============================
df = pd.DataFrame({
    'date': ['2024-01-01', '2024-01-01', '2024-01-02', '2024-01-02'],
    'city': ['Bangkok', 'Phuket', 'Bangkok', 'Phuket'],
    'product': ['A', 'B', 'A', 'B'],
    'sales': [100, 150, 120, 180],
    'quantity': [5, 7, 6, 9]
})

# แปลง date เป็น datetime
df['date'] = pd.to_datetime(df['date'])

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# Pivot Table (มี aggregation)
# ===============================
pivot_table = df.pivot_table(
    values='sales',
    index='city',
    columns='product',
    aggfunc='sum',
    fill_value=0
)

print("\nPivot Table (ยอดขายรวม)")
print(pivot_table)

# ===============================
# Pivot (ไม่มี aggregation)
# ===============================
df_pivot = df.pivot(
    index='date',
    columns='city',
    values='sales'
)

print("\nPivot (sales ตาม date และ city)")
print(df_pivot)

# ===============================
# Melt (Unpivot)
# ===============================
df_melted = pivot_table.reset_index().melt(
    id_vars='city',
    var_name='product',
    value_name='sales'
)

print("\nMelted DataFrame")
print(df_melted)
```
