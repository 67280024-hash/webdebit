```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# สร้างข้อมูล
data = np.random.randn(1000)

# สร้าง figure และ subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Histogram
axes[0, 0].hist(data, bins=30, color="skyblue", edgecolor="black")
axes[0, 0].set_title("Histogram")

# Plot 2: Box plot
axes[0, 1].boxplot(
    data,
    patch_artist=True,
    boxprops=dict(facecolor="skyblue")
)
axes[0, 1].set_title("Box Plot")

# Plot 3: Violin plot (Seaborn)
sns.violinplot(y=data, ax=axes[1, 0], color="skyblue")
axes[1, 0].set_title("Violin Plot")

# Plot 4: KDE plot
sns.kdeplot(data, ax=axes[1, 1], fill=True, color="skyblue")
axes[1, 1].set_title("KDE Plot")

# จัด layout และแสดงผล
plt.tight_layout()
plt.show()
```
