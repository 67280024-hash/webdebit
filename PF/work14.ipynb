```python
import pandas as pd
import numpy as np

# ===============================
# สร้าง DataFrame
# ===============================
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'age': [25, 30, 35],
    'salary': [50000, 60000, 75000]
})

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# เพิ่มคอลัมน์ใหม่
# ===============================
df['bonus'] = 5000
df['total'] = df['salary'] + df['bonus']

# คำนวณจากคอลัมน์อื่น
df['tax'] = df['salary'] * 0.07
df['net_salary'] = df['salary'] - df['tax']

# ===============================
# ใช้ apply กับ function
# ===============================
def categorize_age(age):
    if age < 30:
        return 'Young'
    elif age < 40:
        return 'Middle'
    else:
        return 'Senior'

df['age_group'] = df['age'].apply(categorize_age)

# Lambda function
df['salary_k'] = df['salary'].apply(lambda x: x / 1000)

# Apply หลายคอลัมน์
df['age_salary_ratio'] = df.apply(
    lambda row: row['age'] / row['salary'] * 1000,
    axis=1
)

# ===============================
# Insert คอลัมน์
# ===============================
df.insert(2, 'city', 'Bangkok')

print("\nหลังเพิ่มและคำนวณคอลัมน์")
print(df)

# ===============================
# ลบคอลัมน์ (ไม่ใช้ inplace)
# ===============================
df = df.drop(columns=['bonus', 'tax', 'total', 'age_group', 'salary_k'])

# ===============================
# เลือกเฉพาะคอลัมน์ที่ต้องการ
# ===============================
df = df[['name', 'age', 'salary', 'city']]

print("\nข้อมูลสุดท้าย")
print(df)
```
