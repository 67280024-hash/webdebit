```python
import pandas as pd

# ===============================
# ตัวอย่าง Concatenate
# ===============================
df1 = pd.DataFrame({'A': ['A0', 'A1', 'A2']})
df2 = pd.DataFrame({'A': ['A3', 'A4', 'A5']})

df_concat = pd.concat([df1, df2], ignore_index=True)

print("ผลลัพธ์จาก pd.concat()")
print(df_concat)

# ===============================
# ตัวอย่าง Merge (SQL-style join)
# ===============================
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4],
    'name': ['Alice', 'Bob', 'Charlie', 'David']
})

orders = pd.DataFrame({
    'order_id': [101, 102, 103],
    'customer_id': [1, 2, 1],
    'amount': [100, 200, 150]
})

df_merge = pd.merge(
    customers,
    orders,
    on='customer_id',
    how='outer',
    indicator=True
)

print("\nผลลัพธ์จาก pd.merge() แบบ Outer Join")
print(df_merge)
```
