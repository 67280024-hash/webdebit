```python
import pandas as pd
import numpy as np

# ===============================
# ข้อมูลตัวอย่าง (มี missing values)
# ===============================
data = {
    'name': ['Alice', 'Bob', None, 'David', 'Eve'],
    'age': [25, 30, 35, None, 32],
    'city': ['Bangkok', None, 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, np.nan, 55000, 65000]
}

df = pd.DataFrame(data)
print(df)

# ===============================
# 1. ลบข้อมูล (Drop)
# ===============================

# ลบแถวที่มี missing values อย่างน้อย 1 ค่า (ค่า default)
df_dropped = df.dropna()
print(df_dropped)

# ลบแถวที่ missing ทุกคอลัมน์
df_dropped = df.dropna(how='all')
print(df_dropped)

# ลบแถวที่มี missing อย่างน้อย 1 ค่า (เหมือนค่า default)
df_dropped = df.dropna(how='any')
print(df_dropped)

# ลบแถวที่มี missing ในคอลัมน์ที่กำหนด
df_dropped = df.dropna(subset=['age', 'salary'])
print(df_dropped)

# ลบคอลัมน์ที่มี missing values
df_dropped = df.dropna(axis=1)
print(df_dropped)

# threshold: เก็บแถวที่มีค่าอย่างน้อย 3 ค่า (ไม่ใช่ missing)
df_dropped = df.dropna(thresh=3)
print(df_dropped)
```
