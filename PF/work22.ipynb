```python
import numpy as np
import matplotlib.pyplot as plt

# สร้างข้อมูล time series
days = np.arange(365)
sales = np.random.randint(200, 500, size=365)
costs = np.random.randint(100, 300, size=365)
profit = sales - costs

# สร้าง subplots แนวตั้ง
fig, axes = plt.subplots(3, 1, figsize=(12, 8), sharex=True)

# Sales
axes[0].plot(days, sales, label="Sales")
axes[0].legend()
axes[0].set_ylabel("Sales")

# Costs
axes[1].plot(days, costs, label="Costs")
axes[1].legend()
axes[1].set_ylabel("Costs")

# Profit
axes[2].plot(days, profit, label="Profit")
axes[2].legend()
axes[2].set_ylabel("Profit")
axes[2].set_xlabel("Day")

# Overall title
fig.suptitle("Company Performance Over 365 Days", fontsize=16)
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```
