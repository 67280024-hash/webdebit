```python
import pandas as pd
import numpy as np

# ===============================
# 2.1 Missing Values (ค่าที่หายไป)
# ตรวจสอบ Missing Values
# ===============================

# สร้างข้อมูลตัวอย่างที่มี missing values
data = {
    'name': ['Alice', 'Bob', None, 'David', 'Eve'],
    'age': [25, 30, 35, None, 32],
    'city': ['Bangkok', None, 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, np.nan, 55000, 65000]
}

df = pd.DataFrame(data)
print(df)

# -------------------------------
# ตรวจสอบ missing values
# -------------------------------

print(df.isna())               # หรือ df.isnull() → True/False ทุกเซลล์
print(df.isna().sum())         # นับจำนวน missing ต่อคอลัมน์
print(df.isna().sum().sum())   # นับ missing ทั้งหมด

# -------------------------------
# เปอร์เซ็นต์ของ missing values
# -------------------------------
print(df.isna().sum() / len(df) * 100)

# -------------------------------
# แถวที่มี missing values
# -------------------------------
print(df[df.isna().any(axis=1)])

# -------------------------------
# คอลัมน์ที่มี missing values
# -------------------------------
print(df.columns[df.isna().any()])
```
