```python
import numpy as np

# ===============================
# 4.1 Temperature Data Analysis
# ===============================

temperatures = np.array([
    32, 34, 31, 29, 30, 33, 35, 28, 36, 30,
    31, 32, 34, 33, 31, 29, 30, 32, 33, 31
])

# a) จำนวนวันที่อุณหภูมิมากกว่า 32
count_hot_days = (temperatures > 32).sum()

# b) ค่าเฉลี่ยของวันที่อุณหภูมิมากกว่า 32
avg_hot_days = temperatures[temperatures > 32].mean()

# c) วันที่อุณหภูมิต่ำกว่า 30 หรือสูงกว่า 34
extreme_days = temperatures[(temperatures < 30) | (temperatures > 34)]

# d) แทนที่อุณหภูมิต่ำกว่า 30 ด้วย 30
adjusted_temp = temperatures.copy()
adjusted_temp[adjusted_temp < 30] = 30

# e) indices ของวันที่อุณหภูมิสูงสุด
max_temp_indices = np.where(temperatures == temperatures.max())[0]

print("=== ข้อ 4.1 ===")
print("จำนวนวันที่อุณหภูมิมากกว่า 32:", count_hot_days)
print("ค่าเฉลี่ยอุณหภูมิ (>32):", avg_hot_days)
print("วันที่อุณหภูมิต่ำกว่า 30 หรือสูงกว่า 34:", extreme_days)
print("อุณหภูมิหลังปรับค่า:", adjusted_temp)
print("indices ของวันที่อุณหภูมิสูงสุด:", max_temp_indices)


# ===============================
# 4.2 Sales Data Filtering
# ===============================

sales = np.array([
    120, 85, 200, 150, 95, 180, 210, 70, 190, 100,
    160, 140, 110, 185, 205, 90, 175, 155, 125, 195
])

# a) ยอดขายระหว่าง 100 - 150
mid_sales = sales[(sales >= 100) & (sales <= 150)]

# b) จำนวนวันที่ขายได้มากกว่า 150
count_high_sales = np.count_nonzero(sales > 150)

# c) วันที่ขายได้น้อยกว่า 100 และแทนที่ด้วย 100
fixed_sales = sales.copy()
fixed_sales[fixed_sales < 100] = 100

# d) หา outliers (ห่างจากค่าเฉลี่ยมากกว่า 2 std)
mean_sales = sales.mean()
std_sales = sales.std()
outliers = sales[np.abs(sales - mean_sales) > 2 * std_sales]

# e) แบ่งยอดขายเป็น 3 กลุ่ม
low_sales = sales[sales < 100]
medium_sales = sales[(sales >= 100) & (sales <= 150)]
high_sales = sales[sales > 150]

print("\n=== ข้อ 4.2 ===")
print("ยอดขายระหว่าง 100-150:", mid_sales)
print("จำนวนวันที่ขายได้มากกว่า 150:", count_high_sales)
print("ยอดขายหลังแทนค่าต่ำกว่า 100:", fixed_sales)
print("Outliers:", outliers)
print("Low Sales (<100):", low_sales)
print("Medium Sales (100-150):", medium_sales)
print("High Sales (>150):", high_sales)
```
