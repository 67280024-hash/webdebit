```python
import pandas as pd
import numpy as np

# สร้างจาก dictionary
data = {
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'age': [25, 30, 35, 28, 32],
    'city': ['Bangkok', 'Chiang Mai', 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, 75000, 55000, 65000]
}

df = pd.DataFrame(data)
print(df)

# สร้างจาก list of dictionaries
data = [
    {'name': 'Alice', 'age': 25, 'city': 'Bangkok'},
    {'name': 'Bob', 'age': 30, 'city': 'Chiang Mai'},
    {'name': 'Charlie', 'age': 35, 'city': 'Phuket'}
]
df = pd.DataFrame(data)
print(df)

# สร้างจาก NumPy array
arr = np.random.randn(5, 3)
df = pd.DataFrame(arr, columns=['A', 'B', 'C'])
print(df)

# สร้างพร้อม custom index (แก้จำนวน index ให้ตรงกับจำนวนแถว)
df = pd.DataFrame(
    data,
    index=['emp1', 'emp2', 'emp3']
)
print(df)
```
