```python
import pandas as pd

# ===============================
# 2.2 Duplicates (ข้อมูลซ้ำ)
# ===============================

# สร้างข้อมูลที่มีค่าซ้ำ
data = {
    'name': ['Alice', 'Bob', 'Alice', 'David', 'Bob'],
    'age': [25, 30, 25, 28, 30],
    'city': ['Bangkok', 'Phuket', 'Bangkok', 'Pattaya', 'Phuket']
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

# -------------------------------
# ตรวจสอบแถวที่ซ้ำ
# -------------------------------

print("\ndf.duplicated():")
print(df.duplicated())           # True/False สำหรับแต่ละแถว

print("\nNumber of duplicated rows:")
print(df.duplicated().sum())     # นับจำนวนแถวที่ซ้ำ

# แสดงแถวที่ซ้ำ (ค่า True)
print("\nDuplicated rows:")
print(df[df.duplicated()])

# -------------------------------
# ตรวจสอบการซ้ำในคอลัมน์เฉพาะ
# -------------------------------
print("\nDuplicated by 'name':")
print(df.duplicated(subset=['name']))

# -------------------------------
# ตัวเลือก keep
# -------------------------------

print("\nKeep last occurrence:")
print(df.duplicated(keep='last'))

print("\nMark all duplicates:")
print(df.duplicated(keep=False))

# -------------------------------
# ลบแถวที่ซ้ำ
# -------------------------------

# ลบแถวที่ซ้ำ (default เก็บตัวแรก)
df_unique = df.drop_duplicates()
print("\nAfter drop_duplicates() (keep first):")
print(df_unique)

# ลบตามคอลัมน์เฉพาะ
df_unique = df.drop_duplicates(subset=['name'])
print("\nAfter drop_duplicates(subset=['name']):")
print(df_unique)

# เก็บตัวสุดท้าย
df_unique = df.drop_duplicates(keep='last')
print("\nAfter drop_duplicates(keep='last'):")
print(df_unique)
```
