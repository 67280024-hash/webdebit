```python
import pandas as pd

# ===============================
# ข้อมูลตัวอย่าง
# ===============================
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 35, None],
    'salary': [50000, 60000, 75000, 55000],
    'city': ['Bangkok', 'Phuket', 'Bangkok', 'Bangkok']
})

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# 1) เรียงตามอายุ (น้อย → มาก)
# ===============================
df_sorted_age = df.sort_values('age')
print("\nเรียงตาม age (น้อย → มาก)")
print(df_sorted_age)

# ===============================
# 2) เรียงตามอายุ (มาก → น้อย)
# ===============================
df_sorted_age_desc = df.sort_values('age', ascending=False)
print("\nเรียงตาม age (มาก → น้อย)")
print(df_sorted_age_desc)

# ===============================
# 3) เรียงหลายคอลัมน์
# city ↑ และ age ↓
# ===============================
df_sorted_multi = df.sort_values(
    ['city', 'age'],
    ascending=[True, False]
)
print("\nเรียงตาม city (↑) และ age (↓)")
print(df_sorted_multi)

# ===============================
# 4) เรียงตาม index
# ===============================
df_sorted_index = df.sort_index()
print("\nเรียงตาม index")
print(df_sorted_index)

# ===============================
# 5) เรียงตาม salary (มาก → น้อย)
# ===============================
df_sorted_salary = df.sort_values('salary', ascending=False)
print("\nเรียงตาม salary (มาก → น้อย)")
print(df_sorted_salary)

# ===============================
# 6) จัดการ NaN ในการเรียง
# ===============================
df_sorted_na = df.sort_values('age', na_position='first')
print("\nเรียง age โดยให้ NaN อยู่ก่อน")
print(df_sorted_na)
```
