```python
import pandas as pd
import numpy as np

# ===============================
# สร้างช่วงวันที่
# ===============================
dates = pd.date_range(start='2024-01-01', periods=10, freq='D')

df = pd.DataFrame({
    'date': dates,
    'value': np.random.randn(10)
})

print("ข้อมูลเริ่มต้น")
print(df)

# ===============================
# แยกข้อมูลวันที่
# ===============================
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day_name'] = df['date'].dt.day_name()

print("\nหลังแยกปี เดือน และวัน")
print(df)

# ===============================
# ตั้ง date เป็น index
# ===============================
df = df.set_index('date')

# ===============================
# คำนวณจำนวนวันตั้งแต่วันแรก
# ===============================
df['days_since_start'] = (df.index - df.index.min()).days

print("\nDataFrame หลังตั้ง index และคำนวณวัน")
print(df)
```
