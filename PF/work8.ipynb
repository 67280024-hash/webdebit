```python
import pandas as pd
import numpy as np

# ===============================
# ข้อมูลตัวอย่าง
# ===============================
data = {
    'name': ['Alice', 'Bob', None, 'David', 'Eve'],
    'age': [25, 30, 35, None, 32],
    'city': ['Bangkok', None, 'Phuket', 'Bangkok', 'Pattaya'],
    'salary': [50000, 60000, np.nan, 55000, 65000]
}

df = pd.DataFrame(data)
print(df)

# ===============================
# เติม Missing Values
# ===============================

# 1) เติมด้วยค่าเฉพาะ
df_filled = df.fillna(0)
df_filled = df.fillna('Unknown')

# 2) เติมแต่ละคอลัมน์ต่างกัน
df_filled = df.fillna({
    'age': df['age'].mean(),
    'city': 'Unknown',
    'salary': df['salary'].median()
})

# 3) Forward fill (ใช้ค่าก่อนหน้า)
df_filled = df.fillna(method='ffill')  # หรือ 'pad'

# 4) Backward fill (ใช้ค่าถัดไป)
df_filled = df.fillna(method='bfill')

# 5) Interpolate (เฉพาะตัวเลข)
df_filled = df.interpolate()

# ===============================
# เติมด้วย mean / median / mode (แก้ inplace)
# ===============================
df['age'].fillna(df['age'].mean(), inplace=True)
df['salary'].fillna(df['salary'].median(), inplace=True)
df['city'].fillna(df['city'].mode()[0], inplace=True)

print(df)
```
